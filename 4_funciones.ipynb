### <font color= "darkblue">**Producto cardinal y funciones**</font>

#### **Definición 2.1.1 (Producto cardinal)**
Sean $A=\{a_1,a_2\}$ y $B=\{b_1,b_2\}$ dos conjuntos. Siempre es posible definir el producto cardinal $A \times B$, como el conjunto de **pares ordenados** $(x,y)$, donde $x \in A \land y \in B$. Luego, por extensión:<br><br>
\begin{equation}
    A \times B=\{(a_1,b_1),(a_1,b_2),(a_2,b_1),(a_2,b_2)\}
\end{equation}
<p align= "right">=^.^=</p>

#### **Definición 2.1.2 (Función)**
Subconjunto del producto cardinal en el cual todos los elementos del primer conjunto ($A$ en la definición 2.1.1) están representados una única vez.
<p align= "right">=^.^=</p>

#### **Ejemplo 2.1.3 (Subconjuntos de $A\times B$)**
Consideremos el caso de la definición 2.1.1, y definamos si los siguientes subconjuntos de $A\times B$ son funciones:<br>
(1) $\{(a_1,b_1),(a_2,b_2)\}$ $\rightarrow$ **Función**<br>
(2) $\{(a_1,b_1),(a_2,b_1)\}$ $\rightarrow$ **Función**, aunque $b_1 \in B$ esté representado más de una vez, y $b_2 \in B$ no esté representado. <br>
(3) $\{(a_1,b_2)\}$ $\rightarrow$ **<span style="color: red;">No</span> es Función**, ya que $a_2 \in A$ no está representado. <br>
(4) $\{(a_1,b_1),(a_1,b_2),(a_2,b_1)\}$ $\rightarrow$ **<span style="color: red;">No</span> es Función**, ya que $a_1 \in A$ está representado más de una vez. <br>
<p align= "right">=^.^=</p>

#### **Definición 2.1.5 (Dominio y Codominio)**
Sea $f \subseteq A \times B$ una función. Los conjuntos A y B se denominan respectivamente **dominio de f** ($Dom(f)$) y **codominio de f** ($Cod(f)$). En general, para especificar el nombre, dominio y codominio de una función, se usa la notación:<br><br>
\begin{equation}
    f:A \rightarrow B
\end{equation}<br>
Y decimos que $f$ es una función de $A$ en $B$.
<p align= "right">=^.^=</p>

#### **Definición 2.1.6 (Imagen)**
Sea $f \subseteq A \times B$ una función. La imagen de $f$ es el subconjunto de $B$ de los valores que están representados en $f$.
<p align= "right">=^.^=</p>

### <font color= "darkblue">**Funciones de $\mathbb{R}$ en $\mathbb{R}$**</font>

Los conjuntos $A$ y $B$ de la sección, contenían a los sumo 3 elementos. Sin embargo, existen conjuntos con más elementos, incluso infinitos, como es el caso de los números reales $\mathbb{R}$. En general, cuando pensamos en una función, se nos viene a la mente una en la que dominio y codominio corresponden a $\mathbb{R}$:<br>
\begin{equation}
    f:\mathbb{R} \rightarrow \mathbb{R}
\end{equation}

Algunos ejemplos son:<br>
$f_1(x)=2x+4$<br>
$f_2(x)=x^2-1$<br>
$f_3(x)=1.023 \cdot log{(x)}$<br><br>
Sin embargo, ***¿qué significan estas ecuaciones en el contexto de la definición de función 2.1.2?***<br><br>
Tomemos el caso de $f(x)=2x+4$:
\begin{align}
    f:\mathbb{R} &\rightarrow \mathbb{R}\\
    x &\rightarrow f(x)=2x+4
\end{align}<br>
Esto significa que $f \subseteq \mathbb{R}\times\mathbb{R}$ (o bien $f \subseteq \mathbb{R}^2$), y que comprende los pares ordenados $(x,y)$ tales que $x \in \mathbb{R}$, $y \in \mathbb{R}$ y $y=2x+4$. Por ejemplo, $(1,6) \in f$ porque $6=2(1)+4$, pero $(0,0) \notin f$ porque $0\neq 2(0)+4=4$.<br>

### <font color= "darkblue">**Funciones de $\mathbb{R}^2$ en $\mathbb{R}$ y sus gráficas**</font>
Así como podemos hacer el producto cardinal $\mathbb{R} \times \mathbb{R} \equiv \mathbb{R}^2$, también es posible obtener el producto cardinal $\mathbb{R}^2 \times \mathbb{R}$, ya que $\mathbb{R}^2$ es, en sí, un conjunto. De este modo, también existen funciones de $\mathbb{R}^2$ en $\mathbb{R}$ (cuyo dominio es $\mathbb{R}^2$ y su codominio es $\mathbb{R}$):<br><br>
\begin{equation}
    g:\mathbb{R}^2 \rightarrow \mathbb{R}
\end{equation}<br>
Algunos ejemplos son:<br><br>
$g_1(x,y)=x+y$<br>
$g_2(x,y)=x^2+y^2$<br>
$g_3(x,y)=x^2-y^2$<br>
$g_4(x,y)=-\frac{1}{\sqrt{x^2+y^2}}$<br><br>
Puesto que los elementos de una función de $\mathbb{R}^2$ en $\mathbb{R}$ están en $\mathbb{R}^3$, podemos representarlos en un sistema de coordenadas cartesiano tridimensional, y así construir una gráfica de la función. <br>
Graficamos $g(x,y)=x^2+y^2$ en un sistema de coordenadas cartesiano tridimensional:


```python
# Definimos ahora nuestra función de R2 en R

def g(x,y):
    return x**2+y**2
```

```python
# Una vez definida la función, podemos obtener el valor f(x,y) que le corresponde a algún valor de x en el dominio:

g(3,2)
```

```python
# Graficamos la función en el sistema de coordenadas cartesiano tridimensional

# Definimos los valores de x e y con np.linspace(inicio, final, pasos)
x_val=np.linspace(-1.5,1.5,100)
y_val=np.linspace(-1.5,1.5,100)

# Graficamos con la función plot_r2_r

fx.plot_r2_r(x_val,y_val,g)
```

### <font color= "darkblue">**Curvas de nivel en funciones de $\mathbb{R}^2$ en $\mathbb{R}$**</font>

Tomemos la función:<br><br>
\begin{align}
    g:\mathbb{R}^2 &\rightarrow \mathbb{R}\\
    (x,y) &\rightarrow g(x,y)=x^2+y^2
\end{align}<br><br>
Busquemos todos aquellos puntos del dominio para los cuales $g(x,y)=1$. Entonces, sustituimos:<br><br>
\begin{align}
    g(x,y)&=x^2+y^2 \implies\\
    1&=x^2+y^2 \implies\\
    \sqrt{1}&=\sqrt{x^2+y^2} \implies\\
    1&=\sqrt{x^2+y^2}
\end{align}<br><br>
#### Comentarios
- Es decir, para todos los puntos de $\mathbb{R}^2$ para los cuales $g(x,y)=1$, se cumple la condición: $\sqrt{x^2+y^2}=1$.<br>
- Aplicando el *teorema de Pitágoras* no es difícil mostrar que la distancia $d$ de un punto $(x,y)$ al origen de un plano de coordenadas cartesiano, viene dada por  $r=\sqrt{x^2+y^2}$<br>
- De este modo ***si representamos en un plano cartesiano a todos los puntos de $\mathbb{R}^2$ para los cuales $g(x,y)=1$ , estos forman una circunferencia de radio $r=1$ centrada en el origen***, como se puede ver en las gráficas siguientes.<br>
- Habremos graficado la curva de nivel $g(x,y)=1$:
#### **Definición 2.1.9 (Curva de nivel en una función de $\mathbb{R}^2$ en $\mathbb{R}$)**
Sea $g:\mathbb{R}^2 \rightarrow \mathbb{R}$, $z \in \mathbb{R}$. La curva de nivel $g(x,y)=z$ es el conjunto de elementos de $\mathbb{R}^2$ (o más rigurosamente, del dominio de $g$), tales que $g(x,y)=z$.
<p align= "right">=^.^=</p>

### <font color= "darkblue">**Funciones de $\mathbb{R}^3$ en $\mathbb{R}$ y sus gráficas**</font>
Ya habrás notado que existe flexibilidad en la elección del dominio y codominio de una función, para así conseguir funciones de $\mathbb{R}^n$ en $\mathbb{R}^m$, con $n,m \in \mathbb{N}$. Para nuestros fines del curso, nos centraremos en las funciones de $\mathbb{R}^3$ en $\mathbb{R}$:<br><br>
\begin{equation}
    F:\mathbb{R}^3 \rightarrow \mathbb{R}
\end{equation}<br>
Algunos ejemplos son:<br><br>
$F_1(x,y,z)=x^2+y^2+z^2$<br>
$F_2(x,y,z)=x^2+y^2+z^2$<br>
En este caso, los elementos de la función pertenecen a $\mathbb{R}^4$, y no podemos representarlos en un sistema de coordenadas cartesiano. Sin embargo, los elementos del dominio pertenecen a $\mathbb{R}^3$, y análogamente a lo que hicimos con las curvas de nivel, podemos construir **superficies de nivel** para representar a todos los elementos del dominio tales que $F(x,y,z)=C$.<br><br>
#### **Definición 2.1.10 (Superficie de nivel en una función de $\mathbb{R}^3$ en $\mathbb{R}$)**
Sea $F:\mathbb{R}^3 \rightarrow \mathbb{R}$, $C \in \mathbb{R}$. La superficie de nivel $F(x,y,z)=C$ es el conjunto de elementos de $\mathbb{R}^3$ (o más rigurosamente, del dominio de $F$), tales que $F(x,y,z)=C$.
<p align= "right">=^.^=</p>

Consideremos la función:<br><br>
\begin{align}
    g:\mathbb{R}^3 &\rightarrow \mathbb{R}\\
    (x,y) &\rightarrow g(x,y)=x^2+y^2+z^2
\end{align}<br><br>
Y grafiquemos algunas superficies de nivel:

```python
# Definimos la función de R3 en R

def F(x,y,z):
    return x**2+y**2+z**2
```

```python
#Definimos los valores de x,y,z

x_val = np.linspace(-3, 3, 30)
y_val = np.linspace(-3, 3, 30)
z_val = np.linspace(-3, 3, 30)

# Definimos los niveles de las superficies
niveles = np.linspace(0, 10, 20)

# Graficamos isosuperficies en forma dinámica con la función isosurface
fx.isosurface(x_val,y_val,z_val,F,niveles)
```
